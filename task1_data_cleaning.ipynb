```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
filename = next(iter(uploaded))
df = pd.read_csv(filename)
original_rows = len(df)
print("Rows and columns:", df.shape)
display(df.head())
df.info()
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Example: "Date Added" becomes "date_added"
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)
# Remove unnecessary spaces from text values
text_columns = df.select_dtypes(include=["object", "string"]).columns
for column in text_columns:
    df[column] = df[column].str.strip()
# Treat empty text as missing
df = df.replace(r"^\s*$", pd.NA, regex=True)
```

```python
duplicates_removed = int(df.duplicated().sum())
df = df.drop_duplicates().copy()
print("Duplicate rows removed:", duplicates_removed)
```

```python
print(df.isnull().sum())
```

```python
for column in ["director", "cast", "country"]:
    if column in df.columns:
        df[column] = df[column].fillna("Unknown")
```

```python
if "date_added" in df.columns:
    original_dates = df["date_added"].copy()
    df["date_added"] = pd.to_datetime(
        df["date_added"],
        format="%B %d, %Y",
        errors="coerce"
    )
    invalid_dates = original_dates.notna() & df["date_added"].isna()
    print("Dates that could not be converted:", invalid_dates.sum())
    display(original_dates[invalid_dates])
if "release_year" in df.columns:
    df["release_year"] = pd.to_numeric(
        df["release_year"], errors="coerce"
    ).astype("Int64")
```

```python
print("Final shape:", df.shape)
print("\nRemaining missing values:")
print(df.isnull().sum())
print("\nRemaining duplicates:", df.duplicated().sum())
df.info()
display(df.head())
```

```python
df.to_csv(
    "cleaned_dataset.csv",
    index=False,
    date_format="%d-%m-%Y"
)
files.download("cleaned_dataset.csv")
```
